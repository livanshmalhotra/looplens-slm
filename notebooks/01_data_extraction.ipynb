```python
import pandas as pd
import json
import os

# SQL Extraction Query example for LoopLens Database (public schema)
SQL_QUERY = """
SELECT 
    i.line_item_id,
    i.line_item_description,
    i.po_line_desc,
    i.spend_category,
    i.vendor_name,
    i.unit_price_local,
    i.total_amount_local,
    i.currency,
    i.gl_account,
    c.level1 AS level_1,
    c.level2 AS level_2,
    c.level3 AS level_3,
    c.category_name,
    c.unspsc_code,
    c.unspsc_title,
    s.supplier_id,
    s.supplier_name AS normalized_vendor_name
FROM public.fact_invoice i
LEFT JOIN public.dim_category c ON i.category_id = c.category_id
LEFT JOIN public.dim_supplier s ON i.supplier_id = s.supplier_id;
"""
print("Query template ready for extraction.")
```
